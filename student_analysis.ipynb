```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("student.csv.csv")

# Show first rows
print(df.head())

# Dataset information
print(df.info())

# Average study time
print("Average Study Time:", df["StudyTimeWeekly"].mean())

# Average absences
print("Average Absences:", df["Absences"].mean())

# Students with highest study time
top_students = df.sort_values(by="StudyTimeWeekly", ascending=False)
print(top_students.head())

# Bar graph
values = [
    df["StudyTimeWeekly"].mean(),
    df["Absences"].mean()
]

labels = ["Study Time", "Absences"]

plt.figure(figsize=(6,4))
plt.bar(labels, values)
plt.xlabel("Categories")
plt.ylabel("Average Values")
plt.title("Average Study Time vs Absences")
plt.show()

# Scatter plot
plt.figure(figsize=(6,4))
plt.scatter(df["StudyTimeWeekly"], df["Absences"])
plt.xlabel("Study Time Weekly")
plt.ylabel("Absences")
plt.title("Study Time vs Absences")
plt.show()
```
